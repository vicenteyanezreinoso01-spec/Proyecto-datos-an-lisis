# Ejercicio 6

```python
# Ejercicio 6 (mas complejo):
#"Necesito un análisis de rentabilidad. Quiero que excluyamos los pedidos chicos (ventas de 50 dólares o menos, no son representativos). Con los pedidos que queden, arma un resumen por Región y Categoría que muestre las ventas totales y la ganancia total de cada combinación. Además, agrégale una columna de margen de ganancia (ganancia dividida en venta), para saber qué combinaciones son más rentables, no solo cuáles venden más. Ordénalo mostrando primero las combinaciones más rentables."
# 1. Primero filtra orders_df para quedarte solo con pedidos donde Sales > 50 (usa .loc)
# 2. Con ese resultado filtrado, agrupa por Region y Category
# 3. Calcula la suma de Sales como "venta_total" y la suma de Profit como "ganancia_total"
# 4. Crea una columna nueva "margen" = ganancia_total / venta_total (columna calculada DESPUES del groupby)
# 5. Ordena de mayor a menor margenn



ventas_mayores_50 = orders_df.loc[orders_df["Sales"] > 50]

resumen = ventas_mayores_50.groupby(["Region","Category"]).agg(
                                  ventas_totales=("Sales","sum"),
                                 ganancia_total=("Profit","sum"))

resumen["margen"] = resumen["ganancia_total"] / resumen["ventas_totales"]
resumen_ordenado = resumen.sort_values(by="margen", ascending=False)

resumen_ordenado.head()
```

# Ejercicio 7

```python
#Ejercicio 7
#"El equipo comercial quiere saber si el descuento se está usausandondo de forma razonable. Para cada Segmento de cliente, necesito saber: cuántos pedidos tuvo, el descuento promedio aplicado, y la ganancia promedio. Pero además, quiero ver esto separado en dos grupos: pedidos donde SÍ hubo descuento (Discount > 0) versus donde NO hubo descuento — para comparar si dar descuento realmente afecta la ganancia."


pedidos_con_descuento = orders_df.loc[orders_df["Discount"] > 0]
pedidos_sin_descuento = orders_df.loc[orders_df["Discount"] == 0]


segmento_sin_descuento = pedidos_sin_descuento.groupby("Segment").agg(
                  n_pedido=("Segment","count"),
                  descuento_promedio=("Discount","mean"),
                  ganancia_promedio=("Profit","mean"))

segmento_con_descuento = pedidos_con_descuento.groupby("Segment").agg(
                  n_pedido=("Segment","count"),
                  descuento_promedio=("Discount","mean"),
                  ganancia_promedio=("Profit","mean"))

segmento_sin_descuento["grupo"] = "Sin descuento"
segmento_con_descuento["grupo"] = "Con descuento"

new_data = pd.concat([segmento_sin_descuento, segmento_con_descuento])
new_data.head(100)
```

```python
# EjercicioEjercicio 9:
# "Necesito identificar los pedidos 'problematicos': aquellos con Discount mayor a 0.3
# Y ganancia negativa (Profit < 0) al mismo tiempo. Marca estos pedidos con una columna
# 'Problema' usando np.where ('Si'/'No'). Luego agrupa por Region y esa columna, mostrando
# cuantos pedidos hay y la suma de Sales perdida en cada grupo. Finalmente, quedate solo
# con las filas donde Problema sea 'Si'."

orders_df["Problematicos"] = np.where((orders_df["Discount"] > 0.3) &
                                 (orders_df["Profit"] < 0),"Si","No")

agrupacion_region_problematico = orders_df.groupby(["Region","Problematicos"]). agg(
                              cantidad_pedidos=("Problematicos","count"),
                              Ventas_total=("Sales","sum"))

agrupacion_region_problematico = agrupacion_region_problematico.reset_index()

agrupacion_region_problematico = agrupacion_region_problematico.loc[
            agrupacion_region_problematico["Problematicos"] == "Si"]


agrupacion_region_problematico.head()
```

# Ejercicio 10

```python
# Ejercicio 10:
#"Necesito un análisis de riesgo de descuento. Clasifica cada pedido en 3 niveles según el Discount: 'Bajo' (0.15 o menos), 'Medio' (más de 0.15 hasta 0.3), 'Alto' (más de 0.3) — usa np.select. Agrupa por Region y ese nivel de riesgo. Calcula: cantidad de pedidos, venta total, ganancia total, y un margen (ganancia dividida en venta, calculado después de agrupar). Finalmente, quédate solo con las combinaciones de riesgo 'Alto' que tengan margen negativo, y ordénalas de la más negativa a la menos negativa."

condiciones = [(orders_df["Discount"] <= 0.15),
                ((orders_df["Discount"] > 0.15) & (orders_df["Discount"] <= 0.3)),
                   (orders_df["Discount"] > 0.3)]

resultados = ["Bajo","Medio","Alto"]

orders_df["Risk"] = np.select(condiciones, resultados ,default="Otro")

data_regiones_risk = orders_df.groupby(["Region","Risk"]).agg(
              cantidad_pedido=("Region","count"),
              vental_total=("Sales","sum"),
              ganancia_total=("Profit","sum"))

data_regiones_risk = data_regiones_risk.reset_index()
data_regiones_risk["margen"] = data_regiones_risk["ganancia_total"] / data_regiones_risk["vental_total"]

data_regiones_risk = data_regiones_risk.loc[
    (data_regiones_risk["Risk"]=="Alto") &
    (data_regiones_risk["margen"] < 0)
]

data_regiones_risk = data_regiones_risk.sort_values(by="margen", ascending=True)

data_regiones_risk.head()
```

```python
#Enunciado — Prueba final:

#"Necesito un reporte de riesgo comercial para la gerencia. Primero, descarta los pedidos con Sales de 20 dólares o menos (son ruido, no aportan). Con lo que quede, clasifica cada pedido en un 'Perfil', evaluado en este orden:

#'Descuento excesivo': Discount mayor a 0.4
#'Perdida sin descuento alto': Profit menor a 0 Y Discount de 0.4 o menos
#'Alto volumen bajo margen': Quantity de 6 o más Y Profit menor a 100
#Todo lo demás: 'Normal'

#Agrupa por Region, Category y Perfil (las tres a la vez). Necesito, por cada combinación: cantidad de pedidos, venta total y ganancia total. Agrega una columna de margen (ganancia entre venta).

#Quédate solo con combinaciones que tengan más de 15 pedidos y que NO sean 'Normal' — no me interesa lo normal, quiero ver los perfiles de riesgo. Ordena mostrando primero el peor margen, y déjame solo el top 10 más crítico."

riesgo_df = orders_df.loc[orders_df["Sales"] > 20].copy()

condiciones = [(riesgo_df["Discount"] > 0.40),
               ((riesgo_df["Discount"] <= 0.40) & (riesgo_df["Profit"] < 0)),
                ((riesgo_df["Quantity"] >= 6) & (riesgo_df["Profit"] < 100))]

clasificacion = ["Descuento excesivo", "Perdida sin descuento alto",
                 "Alto volumen bajo margen"]

riesgo_df["Perfil"] = np.select(condiciones, clasificacion, default="Normal")

riesgo_df = riesgo_df.groupby(["Region","Category","Perfil"]).agg(
              cantidad_pedido=("Profit","count"),
              vental_total=("Sales","sum"),
              ganancia_total=("Profit","sum"))

riesgo_df = riesgo_df.reset_index()

riesgo_df["margen"] = riesgo_df["ganancia_total"] / riesgo_df["vental_total"]

riesgo_df = riesgo_df.loc[(riesgo_df["cantidad_pedido"] > 15) & (riesgo_df["Perfil"] !="Normal")]

riesgo_df.nsmallest(10, "margen")
```

### Encargo — Cierre de filter: Clientes de alto riesgo

De: Riesgo Crediticio
Para: Analista de Datos

Necesito una lista de clientes considerados "alto riesgo" para revisión de sus condiciones
de crédito. Un cliente califica como alto riesgo si cumple **cualquiera** de estos dos
escenarios (no hace falta que cumpla ambos, con uno solo ya califica):

**Escenario A** — las tres condiciones a la vez:
- Ha hecho al menos 8 pedidos
- Su descuento promedio supera 0.2
- Su ganancia total (sumada) es negativa

**Escenario B** — las dos condiciones a la vez:
- Su pedido de mayor venta individual (el máximo de Sales que ha hecho) superó los $5,000
- Y tuvo al menos un pedido con pérdida (Profit negativo)

Con los pedidos de todos los clientes que califiquen por cualquiera de los dos escenarios,
clasifica cada pedido individual como "Pérdida" o "Ganancia" según su Profit.

Finalmente, dame un resumen: para cada cliente calificado, cuántos de sus pedidos cayeron
en "Pérdida" y cuántos en "Ganancia".

No te digo qué herramientas usar, ni en qué orden.

```python
#SOLUCIÓN PROPIA
Customers_risk_condition_A = orders_df.groupby("Customer ID").filter(
    lambda x: (x["Order ID"].nunique() > 7) &
              (x["Discount"].mean() > 0.2) &
              (x["Profit"].sum() < 0))

Customers_risk_condition_B = orders_df.groupby("Customer ID").filter(
    lambda x: (x["Sales"].max() > 5000) &
    (x["Profit"].min() < 0))

customer_risk = pd.concat([Customers_risk_condition_A, Customers_risk_condition_B], ignore_index=True).drop_duplicates()

customer_risk["Clasificacion_profit"] = np.where(customer_risk["Profit"] > 0,"Ganancia", "Perdida" )


resumen = customer_risk.pivot_table (
          index= "Customer ID",
          columns= "Clasificacion_profit",
          values="Sales",
          aggfunc="count",
          margins=True,
          margins_name="Total")

resumen
```

```python
#SOLUCION B
clientes_riesgo = orders_df.groupby("Customer ID").filter(
    lambda x: (
        (x["Order ID"].nunique() >= 8) &
        (x["Discount"].mean() > 0.2) &
        (x["Profit"].sum() < 0)
    ) | (
        (x["Sales"].max() > 5000) &
        (x["Profit"].min() < 0)))

clientes_riesgo["Clasificacion_profit"] = np.where(
    clientes_riesgo["Profit"] > 0, "Ganancia", "Pérdida")

resumen = clientes_riesgo.pivot_table(
    index="Customer ID",
    columns="Clasificacion_profit",
    values="Sales",
    aggfunc="count",
    fill_value=0,
    margins=True,
    margins_name="Total")

resumen
```
